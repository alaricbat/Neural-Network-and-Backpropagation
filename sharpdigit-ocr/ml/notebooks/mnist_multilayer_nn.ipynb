```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

## Import Dataset

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
```

## Processing Data

```python
X_train_flatten = X_train.reshape(X_train.shape[0], -1)
```

```python
X_train_flatten.shape
```

## Train Data

```python
import os
import sys
```

```python
from src.clazz.MultilayerPerceptron import MultilayerPerceptron
```

```python
multilayer = MultilayerPerceptron()
multilayer.fit(X=X_train_flatten, y=y_train)
```

## Predict Data

```python
y_train_pred = multilayer.predict(X_train_flatten)
```

```python
X_test_flatten = X_test.reshape(X_test.shape[0], -1)
y_test_pred = multilayer.predict(X_test_flatten)
```

## Evaluation Prediction

```python
from sklearn.metrics import classification_report

print(classification_report(y_train, y_train_pred))
```
